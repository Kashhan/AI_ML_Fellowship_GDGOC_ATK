## Introduction

Market Basket Analysis is a data mining technique used to identify purchasing patterns in transactional data. It discovers relationships between products frequently bought together.

Association Rule Mining is used to generate rules such as:

If a customer buys Bread → they are likely to buy Butter.

In this notebook we will:
- Load a market basket dataset
- Preprocess transactional data
- Apply Apriori Algorithm
- Generate association rules
- Analyze product relationships

```python
!pip install mlxtend
```

```python
import pandas as pd
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, association_rules
```

```python
dataset = [
    ['Milk', 'Bread', 'Butter'],
    ['Bread', 'Butter'],
    ['Milk', 'Bread'],
    ['Milk', 'Butter'],
    ['Bread', 'Butter', 'Jam'],
    ['Milk', 'Bread', 'Butter', 'Jam'],
    ['Bread'],
    ['Milk', 'Bread', 'Jam'],
    ['Butter', 'Jam'],
    ['Milk', 'Bread', 'Butter']
]
```

```python
te = TransactionEncoder()
te_data = te.fit(dataset).transform(dataset)

df = pd.DataFrame(te_data, columns=te.columns_)

df
```

## **Apply Apriori Algorithm**
The Apriori algorithm identifies frequent itemsets in transactional datasets using minimum support.

```python
frequent_itemsets = apriori(df, min_support=0.3, use_colnames=True)

frequent_itemsets
```

## Association Rule Generation
Association rules help identify relationships between products.

```python
rules = association_rules(frequent_itemsets, metric="confidence", min_threshold=0.6)

rules
```

```python
rules[['antecedents','consequents','support','confidence','lift']]
```

## Result Interpretation

From the generated rules we can observe relationships between products.

Example insights:
- Customers who buy Bread often buy Butter.
- Milk and Bread frequently appear together.
- Jam is often purchased with Butter.

Retail stores can use these patterns to improve:
- Product placement
- Promotional bundles
- Cross selling strategies

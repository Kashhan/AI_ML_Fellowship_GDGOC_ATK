# **TASK 1 – Titanic Missing Data**

### Introduction
This task identifies and handles missing values in the Titanic dataset.

Techniques used:
- Median imputation for Age
- Mode imputation for Embarked

Median is robust to skew. Mode suits categorical features. Cabin has excessive missing values.

# **TASK 4: Handling Outliers – Boston Housing Dataset**

### Introduction

This task focuses on detecting and treating outliers in the Boston Housing dataset.

Techniques used:
- Z score method
- IQR method

Z score identifies extreme observations based on standard deviation from the mean.  
The IQR method detects values outside the interquartile range and is more robust for skewed distributions.

```python
import pandas as pd

df = pd.read_csv(
    '/kaggle/input/datasets/arunjangir245/boston-housing-dataset/BostonHousing.csv'
)

df.head()
```

```python
df = df.select_dtypes(include=['number'])
```

```python
import numpy as np

z_scores = np.abs((df - df.mean()) / df.std())
outliers_z = (z_scores > 3).sum()

outliers_z
```

```python
Q1 = df.quantile(0.25)
Q3 = df.quantile(0.75)
IQR = Q3 - Q1

outlier_condition = ((df < (Q1 - 1.5 * IQR)) | (df > (Q3 + 1.5 * IQR)))
outliers_iqr = outlier_condition.sum()

outliers_iqr
```

```python
df_clean = df[~outlier_condition.any(axis=1)]
df_clean.shape
```

### Results & Observations

Z score method detected extreme values beyond 3 standard deviations.

IQR method identified observations outside interquartile boundaries.

Outliers were removed using IQR to ensure robustness against skewed data.

# **TASK 9: Handling Imbalanced Data – Credit Card Fraud Detection**

### Introduction

This task addresses class imbalance in the Credit Card Fraud dataset.

Techniques used:
- SMOTE
- ADASYN
- Random undersampling

Imbalanced data can bias models toward the majority class. Resampling techniques help balance class distribution.

```python
print(os.listdir('/kaggle/input/datasets/organizations/mlg-ulb/creditcardfraud'))
```

```python
import pandas as pd

df = pd.read_csv(
    '/kaggle/input/datasets/organizations/mlg-ulb/creditcardfraud/creditcard.csv'
)

df.head()
```

```python
X = df.drop('Class', axis=1)
y = df['Class']
```

```python
y.value_counts()
```

```python
from imblearn.over_sampling import SMOTE

smote = SMOTE(random_state=0)
X_sm, y_sm = smote.fit_resample(X, y)

y_sm.value_counts()
```

```python
from imblearn.over_sampling import ADASYN

adasyn = ADASYN(random_state=0)
X_ada, y_ada = adasyn.fit_resample(X, y)

y_ada.value_counts()
```

```python
from imblearn.under_sampling import RandomUnderSampler

rus = RandomUnderSampler(random_state=0)
X_rus, y_rus = rus.fit_resample(X, y)

y_rus.value_counts()
```

### Results & Observations

The dataset showed strong class imbalance with majority non fraud transactions.

SMOTE and ADASYN generated synthetic minority samples to balance classes.

Random undersampling reduced majority class samples to achieve balance.

# **TASK 12: Text Preprocessing – IMDB Dataset of 50K Movie Reviews**

### Introduction

This task performs text preprocessing on the IMDB movie reviews dataset.

Steps applied:
- Lowercasing
- Tokenization
- Stopword removal
- Stemming

Text preprocessing prepares raw text data for machine learning models.

```python
import pandas as pd

df = pd.read_csv(
    '/kaggle/input/datasets/lakshmi25npathi/imdb-dataset-of-50k-movie-reviews/IMDB Dataset.csv'
)

df.head()
```

```python
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
import re

stop_words = set(stopwords.words('english'))
ps = PorterStemmer()
```

```python
ps = PorterStemmer()
stop_words = set(stopwords.words('english'))

def clean_text(text):
    text = text.lower()
    text = re.sub('[^a-zA-Z]', ' ', text)
    words = text.split()
    words = [ps.stem(word) for word in words if word not in stop_words]
    return ' '.join(words)
```

```python
df['clean_review'] = df['review'].apply(clean_text)

df[['review', 'clean_review']].head()
```

### Results & Observations

Text was converted to lowercase and cleaned of non alphabetical characters.

Stopwords were removed to reduce noise.

Stemming reduced words to their root form, improving text normalization.

# **TASK 13: Time Series Preprocessing – UCI Air Quality Dataset**

### Introduction

This task preprocesses time series data in the Air Quality dataset.

Steps applied:
- Handling missing timestamps
- Resampling
- Smoothing using rolling mean

Time series preprocessing ensures consistent intervals and reduces noise.

```python
import pandas as pd

df = pd.read_csv(
    '/kaggle/input/datasets/dakshbhalala/uci-air-quality-dataset/AirQualityUCI.csv',
    sep=','
)

print(df.columns)
print(df.dtypes)
```

```python
df = df.drop(columns=['Unnamed: 15', 'Unnamed: 16'])
```

```python
df['Datetime'] = pd.to_datetime(df['Date'] + ' ' + df['Time'], errors='coerce')

df = df.dropna(subset=['Datetime'])

df = df.set_index('Datetime')
```

```python
import numpy as np

df = df.replace(-200, np.nan)
```

```python
df_numeric = df.select_dtypes(include=['float64', 'int64'])

print(df_numeric.columns)
```

```python
df_resampled = df_numeric.resample('h').mean()
df_resampled.head()
```

```python
df_interpolated = df_resampled.interpolate()
df_interpolated.head()
```

```python
df_interpolated['CO_smoothed'] = df_interpolated['CO(GT)'].rolling(window=3).mean()

df_interpolated[['CO(GT)', 'CO_smoothed']].head()
```

### Results & Observations

Datetime index was created by combining date and time columns.

Data was resampled to hourly frequency.

Missing values were interpolated.

Rolling mean smoothing reduced short term fluctuations in CO levels.

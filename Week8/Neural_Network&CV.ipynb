### **INTRODUCTION TO DEEP LEARNING**

Deep Learning is a subset of machine learning that uses neural networks with multiple layers to learn patterns from data.
It is widely used in image processing, speech recognition, and natural language processing.

### **NEURAL NETWORK FUNDAMENTALS**

A neural network consists of layers of interconnected neurons.
Each neuron receives input, applies weights, adds bias, and passes through an activation function.

Key Components:
- Input Layer
- Hidden Layers
- Output Layer
- Weights and Bias
- Activation Functions

### **PERCEPTRON**

A perceptron is the simplest neural network unit.
It performs a weighted sum of inputs and applies an activation function to produce output.

### **BACKPROPAGATION**

Backpropagation is used to update weights in a neural network.
It calculates error and propagates it backward to minimize loss using gradient descent.

```python
import numpy as np

# Simple forward pass example
inputs = np.array([1, 2, 3])
weights = np.array([0.2, 0.8, -0.5])
bias = 2

output = np.dot(inputs, weights) + bias
print("Output:", output)
```

### **DATA PREPROCESSING (MNIST)**

```python
from tensorflow.keras.datasets import mnist

(X_train, y_train), (X_test, y_test) = mnist.load_data()
```

```python
X_train = X_train / 255.0
X_test = X_test / 255.0
```

### **ANN MODEL**

```python
from tensorflow.keras import models, layers

model_ann = models.Sequential([
    layers.Flatten(input_shape=(28,28)),
    layers.Dense(128, activation='relu'),
    layers.Dense(64, activation='relu'),
    layers.Dense(10, activation='softmax')
])
```

```python
model_ann.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
model_ann.fit(X_train, y_train, epochs=5)
```

```python
model_ann.evaluate(X_test, y_test)
```

### **COMPUTER VISION THEORY**

Computer Vision enables machines to interpret and understand images.

Key Concepts:
- Image classification
- Object detection
- Semantic segmentation

### **CNN THEORY**

CNNs use convolutional layers to extract features from images.

Components:
- Convolutional Layer
- Pooling Layer
- Fully Connected Layer

### **CNN MODEL**

```python
X_train = X_train.reshape(-1,28,28,1)
X_test = X_test.reshape(-1,28,28,1)
```

```python
model_cnn = models.Sequential([
    layers.Conv2D(32, (3,3), activation='relu', input_shape=(28,28,1)),
    layers.MaxPooling2D((2,2)),

    layers.Conv2D(64, (3,3), activation='relu'),
    layers.MaxPooling2D((2,2)),

    layers.Flatten(),
    layers.Dense(128, activation='relu'),
    layers.Dense(10, activation='softmax')
])
```

```python
model_cnn.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
model_cnn.fit(X_train, y_train, epochs=5)
```

```python
model_cnn.evaluate(X_test, y_test)
```

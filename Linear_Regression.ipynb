# **Introduction to Machine Learning**

**Machine Learning** is a branch of Artificial Intelligence where systems learn patterns from data and make predictions without being explicitly programmed. Instead of writing rules manually, we train a model using historical data.

**Types of Machine Learning**
1. **Supervised Learning** Uses labeled data
Example: Predicting house price
2. **Unsupervised Learning** No labels
Example: Customer clustering


**Linear Regression** Linear Regression is a supervised learning algorithm used for predicting continuous values.

It models relationship between:

Independent variable X
Dependent variable Y

Equation:

Y = mX + b

Where
m is slope
b is intercept

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error
```

```python
np.random.seed(42)

X = 10 * np.random.rand(100, 1)

true_slope = 2.5
true_intercept = 4

noise = np.random.randn(100, 1) * 2

y = true_intercept + true_slope * X + noise
y = y.flatten()

print("Number of data points:", len(X))

plt.scatter(X, y)
plt.xlabel("X")
plt.ylabel("y")
plt.title("100 Scattered Data Points")
plt.show()
```

## **Linear Regression Using Scratch**

```python
X_b = np.c_[np.ones((X.shape[0],1)), X]
```

```python
theta = np.linalg.inv(X_b.T.dot(X_b)).dot(X_b.T).dot(y)

b_scratch = theta[0]
m_scratch = theta[1]

print("Scratch Intercept:", b_scratch)
print("Scratch Slope:", m_scratch)
```

```python
y_pred_scratch = m_scratch * X + b_scratch

print("Scratch Predictions:\n", y_pred_scratch.flatten())
```

```python
plt.scatter(X,y,label="Actual")
plt.plot(X,y_pred_scratch,color="red",label="Scratch Line")
plt.legend()
plt.title("Linear Regression From Scratch")
plt.show()
```

## **Linear Regression Using Scikit Learn**

```python
model = LinearRegression()
model.fit(X,y)

b_sklearn = model.intercept_
m_sklearn = model.coef_[0]

print("Sklearn Intercept:", b_sklearn)
print("Sklearn Slope:", m_sklearn)
```

```python
y_pred_sklearn = model.predict(X)

print("Sklearn Predictions:\n", y_pred_sklearn)
```

```python
plt.scatter(X,y,label="Actual")
plt.plot(X,y_pred_sklearn,color="green",label="Sklearn Line")
plt.legend()
plt.title("Linear Regression Using Sklearn")
plt.show()
```

## **Linear Regression Using Gradient Descent**

```python
m = 0
b = 0
learning_rate = 0.001
epochs = 2000
n = len(X)
```

```python
for i in range(epochs):
    y_pred = m * X.flatten() + b

    dm = (-2/n) * np.sum(X.flatten() * (y - y_pred))
    db = (-2/n) * np.sum(y - y_pred)

    m = m - learning_rate * dm
    b = b - learning_rate * db
```

```python
print("Gradient Descent Intercept:", b)
print("Gradient Descent Slope:", m)

y_pred_gd = m * X + b
```

```python
plt.scatter(X,y,label="Actual")
plt.plot(X,y_pred_gd,color="purple",label="Gradient Descent Line")
plt.legend()
plt.title("Linear Regression Using Gradient Descent")
plt.show()
```

## **Compare All Methods**

```python
print("Mean Squared Error Comparison")

mse_scratch = mean_squared_error(y,y_pred_scratch)
mse_sklearn = mean_squared_error(y,y_pred_sklearn)
mse_gd = mean_squared_error(y,y_pred_gd)

print("Scratch MSE:", mse_scratch)
print("Sklearn MSE:", mse_sklearn)
print("Gradient Descent MSE:", mse_gd)
```
